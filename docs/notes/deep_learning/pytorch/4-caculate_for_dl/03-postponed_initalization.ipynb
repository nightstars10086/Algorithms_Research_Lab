```python
import torch 
from torch import nn
from torch.nn.parameter import UninitializedParameter
```

```python
def get_net():
    return nn.Sequential(
        nn.LazyLinear(256),
        nn.ReLU(),
        nn.LazyLinear(10)
    )

net = get_net()

print(net)
```

```python
for name, param in net.named_parameters():
    if isinstance(param, UninitializedParameter):
        print(f'{name} is uninitialized')
    else:
        print(f'{name} is initialized with shape {param.shape}')

    
```

```python
X = torch.rand(2, 20)
```

```python
net(X)
print(net(X).shape)
```
